```python
def ew_covariance(df, lam):
    df = df.dropna()

    X = df.values
    T = len(X)

    # weights: oldest -> newest
    weights = np.array([
        (1 - lam) * lam ** (T - 1 - i)
        for i in range(T)
    ])

    # Normalize weights
    weights = weights / weights.sum()

    mean = np.sum(X * weights[:, None], axis=0)

    X_centered = X - mean

    cov = X_centered.T @ (X_centered * weights[:, None])

    return pd.DataFrame(cov, index=df.columns, columns=df.columns)
```

### Test 2.2: EW Covariance, lambda=0.94

```python
def ew_corr(df, lam):
    # Skip rows containing missing values
    df = df.dropna()

    X = df.values
    T = len(X)

    # EW weights: oldest -> newest
    weights = np.array([
        (1 - lam) * lam ** (T - 1 - i)
        for i in range(T)
    ])

    # Normalize weights
    weights = weights / weights.sum()

    # 1. EW mean
    ew_mean = np.sum(X * weights[:, None], axis=0)

    # 2. Center observations around EW mean
    X_centered = X - ew_mean

    # 3. EW covariance
    ew_cov = X_centered.T @ (X_centered * weights[:, None])

    # 4. EW standard deviations
    ew_std = np.sqrt(np.diag(ew_cov))

    # 5. EW correlation
    ew_corr_matrix = ew_cov / np.outer(ew_std, ew_std)

    return pd.DataFrame(
        ew_corr_matrix,
        index=df.columns,
        columns=df.columns
    )
```

### Test 2.3 Covatiance with EW variance l= 0.97, EW correlation l=0.94

```python
def ew_covariance_mixed(df, lambda_var=0.97, lambda_corr=0.94):
    
    # Remove rows with any missing values
    df = df.dropna()
    X = df.values
    T = len(X)

    # -------------------------
    # EW Variance: lambda = 0.97
    # -------------------------
    w_var = np.array([
        (1 - lambda_var) * lambda_var ** (T - 1 - i)
        for i in range(T)
    ])
    w_var = w_var / w_var.sum()

    # EW mean for variance
    mean_var = np.sum(X * w_var[:, None], axis=0)

    X_centered_var = X - mean_var

    # EW variance
    ew_var = np.sum(
        X_centered_var ** 2 * w_var[:, None],
        axis=0
    )

    # -------------------------
    # EW Correlation: lambda = 0.94
    # -------------------------
    w_corr = np.array([
        (1 - lambda_corr) * lambda_corr ** (T - 1 - i)
        for i in range(T)
    ])
    w_corr = w_corr / w_corr.sum()

    # EW mean for correlation
    mean_corr = np.sum(X * w_corr[:, None], axis=0)

    X_centered_corr = X - mean_corr

    # EW covariance for correlation calculation
    ew_cov_corr = (
        X_centered_corr.T
        @ (X_centered_corr * w_corr[:, None])
    )

    # EW standard deviations
    ew_std_corr = np.sqrt(np.diag(ew_cov_corr))

    # EW correlation
    ew_corr = ew_cov_corr / np.outer(
        ew_std_corr,
        ew_std_corr
    )

    # -------------------------
    # Combine variance + correlation
    # -------------------------
    ew_std_var = np.sqrt(ew_var)

    covariance = ew_corr * np.outer(
        ew_std_var,
        ew_std_var
    )

    return pd.DataFrame(
        covariance,
        index=df.columns,
        columns=df.columns
    )
```

### Test 3.1 near_psd covariance

```python
def near_psd_covariance(cov):
    # Make sure everything is numeric
    cov = cov.astype(float)

    # Check for missing/infinite values
    if not np.isfinite(cov.values).all():
        raise ValueError("Input covariance matrix contains NaN or infinite values.")

    # Standard deviations
    std = np.sqrt(np.diag(cov))

    # Covariance -> correlation
    corr = cov.values / np.outer(std, std)

    # Make symmetric
    corr = (corr + corr.T) / 2

    # Eigenvalue decomposition
    eigenvalues, eigenvectors = np.linalg.eigh(corr)

    # Set negative eigenvalues to zero
    eigenvalues = np.maximum(eigenvalues, 0)

    # Reconstruct correlation matrix
    corr_psd = eigenvectors @ np.diag(eigenvalues) @ eigenvectors.T

    # Normalize diagonal to 1
    d = np.sqrt(np.diag(corr_psd))
    corr_psd = corr_psd / np.outer(d, d)

    # Convert back to covariance
    cov_psd = corr_psd * np.outer(std, std)

    return pd.DataFrame(
        cov_psd,
        index=cov.index,
        columns=cov.columns
    )
```

### Test 3.2 near_psd correlation

```python
def near_psd_correlation(corr):
    """
    Convert a correlation matrix to a near positive
    semi-definite correlation matrix.
    """
    
    # Convert to numpy array
    A = np.asarray(corr, dtype=float)

    # Make symmetric
    A = (A + A.T) / 2

    # Eigenvalue decomposition
    eigenvalues, eigenvectors = np.linalg.eigh(A)

    # Replace negative eigenvalues with 0
    eigenvalues[eigenvalues < 0] = 0

    # Reconstruct
    B = eigenvectors @ np.diag(eigenvalues) @ eigenvectors.T

    # Make symmetric again
    B = (B + B.T) / 2

    # Rescale to correlation matrix (diagonal = 1)
    std = np.sqrt(np.diag(B))
    C = B / np.outer(std, std)

    # Force diagonal to exactly 1
    np.fill_diagonal(C, 1.0)

    return C
```

### Test 3.3 Higham covariance

```python
def higham_psd(matrix, tol=1e-10, max_iter=100):
    A = matrix.values.astype(float)

    # Make symmetric
    A = (A + A.T) / 2

    Y = A.copy()
    delta_S = np.zeros_like(A)

    for _ in range(max_iter):

        Y_old = Y.copy()

        # Projection onto PSD matrices
        R = Y - delta_S

        eigenvalues, eigenvectors = np.linalg.eigh(R)
        eigenvalues = np.maximum(eigenvalues, 0)

        X = eigenvectors @ np.diag(eigenvalues) @ eigenvectors.T

        # Dykstra correction
        delta_S = X - R

        # Projection onto correlation matrices
        Y = X.copy()
        np.fill_diagonal(Y, 1.0)

        # Make symmetric
        Y = (Y + Y.T) / 2

        # Check convergence
        diff = np.linalg.norm(Y - Y_old, 'fro')
        norm = np.linalg.norm(Y, 'fro')

        if diff / norm < tol:
            break

    return pd.DataFrame(
        Y,
        index=matrix.index,
        columns=matrix.columns
    )
```

```python
def higham_covariance(cov):
    # Covariance -> correlation
    std = np.sqrt(np.diag(cov))
    
    corr = cov.values / np.outer(std, std)

    corr = pd.DataFrame(
        corr,
        index=cov.index,
        columns=cov.columns
    )

    # Higham on correlation matrix
    corr_psd = higham_psd(corr)

    # Correlation -> covariance
    cov_psd = corr_psd.values * np.outer(std, std)

    return pd.DataFrame(
        cov_psd,
        index=cov.index,
        columns=cov.columns
    )
```

### Test 4.1 Cholesky PSD

```python
import numpy as np

def chol_psd(matrix):
    A = matrix.values.astype(float)
    n = A.shape[0]

    root = np.zeros((n, n))

    for j in range(n):
        # Calculate diagonal element
        root[j, j] = A[j, j] - np.sum(root[j, :j] ** 2)

        # Handle small numerical errors / zero eigenvalues
        if root[j, j] <= 1e-10:
            root[j, j] = 0.0
        else:
            root[j, j] = np.sqrt(root[j, j])

        # Calculate elements below diagonal
        for i in range(j + 1, n):
            if root[j, j] > 1e-10:
                root[i, j] = (
                    A[i, j] - np.sum(root[i, :j] * root[j, :j])
                ) / root[j, j]
            else:
                root[i, j] = 0.0

    return root
```

```python
def normal_simulation_pd(cov, n_sim=100000):
    mu = np.zeros(cov.shape[0])

    L = np.linalg.cholesky(cov.values)

    Z = np.random.normal(
        size=(n_sim, cov.shape[0])
    )

    X = Z @ L.T + mu

    return X
```

```python
def normal_simulation_psd(cov, n_sim=100000):
    # 0 mean
    mu = np.zeros(cov.shape[0])
    
    # Cholesky root for PSD matrix
    L = chol_psd(cov)
    
    # Standard normal simulations
    Z = np.random.normal(size=(n_sim, cov.shape[0]))
    
    # Simulated observations
    X = Z @ L.T + mu
    
    return X
```

### Test 5.3
Normal Simulation nonPSD Input, 0 mean, near_psd fix - 100,000 simulations, compare input vs output covariance

```python
def normal_simulation_near_psd(cov, n_sim=100000):
    # 1. Fix the non-PSD covariance matrix
    cov_psd = near_psd_covariance(cov)
    
    # 2. 0 mean
    mu = np.zeros(cov.shape[0])
    
    # 3. Cholesky decomposition
    L = np.linalg.cholesky(cov_psd.values)
    
    # 4. Generate standard normal random variables
    Z = np.random.normal(size=(n_sim, cov.shape[0]))
    
    # 5. Generate simulated observations
    X = Z @ L.T + mu
    
    return X, cov_psd
```

### Test 5.5
PCA Simulation, 99% explained, 0 mean - 100,000 simulations compare input vs output covariance

```python
def pca_simulation(cov, n_sim=100000, explained=0.99):
    # 0 mean
    mu = np.zeros(cov.shape[0])
    
    # Eigenvalue decomposition
    eigenvalues, eigenvectors = np.linalg.eigh(cov.values)
    
    # Sort eigenvalues from largest to smallest
    idx = np.argsort(eigenvalues)[::-1]
    eigenvalues = eigenvalues[idx]
    eigenvectors = eigenvectors[:, idx]
    
    # Calculate cumulative explained variance
    explained_ratio = eigenvalues / np.sum(eigenvalues)
    cumulative = np.cumsum(explained_ratio)
    
    # Number of PCs needed for 99% explained variance
    n_components = np.searchsorted(cumulative, explained) + 1
    
    # Keep the selected eigenvalues/eigenvectors
    eigenvalues_pca = eigenvalues[:n_components]
    eigenvectors_pca = eigenvectors[:, :n_components]
    
    # Simulate standard normal PCA factors
    Z = np.random.normal(
        size=(n_sim, n_components)
    )
    
    # Transform PCA factors back to original space
    X = Z @ (
        eigenvectors_pca * np.sqrt(eigenvalues_pca)
    ).T + mu
    
    return X, n_components, cumulative
```

```python
pca_input = dfs["test5_2"]

simulated_data, n_components, cumulative = pca_simulation(
    pca_input,
    100000,
    0.99
)

testout_pca = np.cov(simulated_data, rowvar=False)

testout_pca = pd.DataFrame(
    testout_pca,
    index=pca_input.index,
    columns=pca_input.columns
)

print(testout_pca)
```

### Test 7.3 
T Regression

```python
from scipy.optimize import minimize
from scipy.stats import t

def fit_t_regression(X, y):
    X = np.asarray(X)
    y = np.asarray(y)

    # Add intercept
    X_design = np.column_stack([np.ones(len(X)), X])

    # Initial OLS estimates
    beta_init = np.linalg.lstsq(X_design, y, rcond=None)[0]
    residuals = y - X_design @ beta_init

    sigma_init = np.std(residuals)
    nu_init = 5.0

    # Parameters:
    # alpha, B1, B2, B3, log(sigma), log(nu)
    theta_init = np.concatenate([
        beta_init,
        [np.log(sigma_init), np.log(nu_init)]
    ])

    def neg_log_likelihood(theta):

        beta = theta[:4]
        sigma = np.exp(theta[4])
        nu = np.exp(theta[5])

        residuals = y - X_design @ beta

        ll = np.sum(
            t.logpdf(residuals / sigma, df=nu)
            - np.log(sigma)
        )

        return -ll

    result = minimize(
        neg_log_likelihood,
        theta_init,
        method="BFGS"
    )

    beta = result.x[:4]
    sigma = np.exp(result.x[4])
    nu = np.exp(result.x[5])

    alpha = beta[0]
    B1 = beta[1]
    B2 = beta[2]
    B3 = beta[3]

    return pd.DataFrame({
        "mu": [0],
        "sigma": [sigma],
        "nu": [nu],
        "Alpha": [alpha],
        "B1": [B1],
        "B2": [B2],
        "B3": [B3]
    })
```

### Test 7.4
Calculate AICC on fitted T

```python
def calculate_aicc_t(x):
    x = np.asarray(x)
    x = x[np.isfinite(x)]

    # Fit Student-t distribution
    nu, mu, sigma = t.fit(x)

    # Log likelihood
    log_likelihood = np.sum(
        t.logpdf(x, df=nu, loc=mu, scale=sigma)
    )

    # Number of observations
    n = len(x)

    # Number of estimated parameters
    k = 3   # mu, sigma, nu

    # AIC
    AIC = -2 * log_likelihood + 2 * k

    # AICc
    AICc = AIC + (2 * k * (k + 1)) / (n - k - 1)

    return AICc
```

### Test 7.5 
Fit a Normal Inverse Gaussian by the method of moments. Returns mu, alpha, beta, delta.

```python
def fit_nig_moments(x):
    x = np.asarray(x)
    x = x[np.isfinite(x)]

    # Sample moments
    mean = np.mean(x)
    
    m2 = np.mean((x - mean)**2)
    m3 = np.mean((x - mean)**3)
    m4 = np.mean((x - mean)**4)

    # Variance
    variance = m2

    # Skewness
    skew = m3 / (m2 ** 1.5)

    # Excess kurtosis
    kurt = m4 / (m2 ** 2) - 3

    # beta / alpha
    r_squared = skew**2 / (3 * kurt - 4 * skew**2)

    if r_squared >= 1:
        raise ValueError("Moments do not produce a valid NIG distribution.")

    r = np.sign(skew) * np.sqrt(r_squared)

    # delta * gamma
    q = 9 / (3 * kurt - 4 * skew**2)

    # alpha
    alpha = np.sqrt(
        q / (variance * (1 - r_squared)**2)
    )

    # beta
    beta = r * alpha

    # gamma = sqrt(alpha^2 - beta^2)
    gamma = np.sqrt(alpha**2 - beta**2)

    # delta
    delta = q / gamma

    # mu
    mu = mean - delta * beta / gamma

    return mu, alpha, beta, delta
```

```python
mu, alpha, beta, delta = fit_nig_moments(dfs["test7_5"])

print("mu:", mu)
print("alpha:", alpha)
print("beta:", beta)
print("delta:", delta)
```

### Test 7.6
Fit the same NIG by maximum likelihood. Returns mu, alpha, beta, delta.

```python
import numpy as np
import pandas as pd
from scipy.optimize import minimize
from scipy.special import kv, gammaln


def fit_nig_mle(x):
    x = np.asarray(x, dtype=float)
    x = x[np.isfinite(x)]

    # Initial values from method of moments
    mu0, alpha0, beta0, delta0 = fit_nig_moments(x)

    # Transform:
    # alpha = exp(a)
    # beta  = alpha * tanh(b)
    # delta = exp(d)
    #
    # This guarantees alpha > |beta| and delta > 0.
    theta0 = np.array([
        mu0,
        np.log(alpha0),
        np.arctanh(np.clip(beta0 / alpha0, -0.999999, 0.999999)),
        np.log(delta0)
    ])

    def neg_log_likelihood(theta):
        mu = theta[0]
        alpha = np.exp(theta[1])
        beta = alpha * np.tanh(theta[2])
        delta = np.exp(theta[3])

        gamma = np.sqrt(alpha**2 - beta**2)

        z = np.sqrt(delta**2 + (x - mu)**2)

        # NIG log-pdf
        log_pdf = (
            np.log(alpha)
            + np.log(delta)
            - np.log(np.pi)
            + beta * (x - mu)
            + np.log(kv(1, alpha * z))
            - np.log(z)
            + delta * gamma
        )

        # Protect against numerical problems
        if not np.all(np.isfinite(log_pdf)):
            return 1e100

        return -np.sum(log_pdf)

    result = minimize(
        neg_log_likelihood,
        theta0,
        method="BFGS"
    )

    # Transform parameters back
    mu = result.x[0]
    alpha = np.exp(result.x[1])
    beta = alpha * np.tanh(result.x[2])
    delta = np.exp(result.x[3])

    return pd.DataFrame({
        "mu": [mu],
        "alpha": [alpha],
        "beta": [beta],
        "delta": [delta]
    })
```
